```python
import pandas as pd
import plotly.io as pio
import plotly.express as px

pio.renderers.default = 'notebook_connected'
figIndex = 0
```

```python
path = "/home/other/Wi-Fi-Analizer-Robot/Retrieve_Wi-Fi_Data/data.csv"
wifiDf = pd.read_csv(path)
wifiDf.head()
```

```python
#fig = px.scatter(wifiDf,x="Channel",y="Channel")
wifiDf["Signal Level Mag"] = 10**(wifiDf["Signal Level"]/20)
fig = px.bar(wifiDf, x='SSID', y='Signal Level Mag', color='Frequency', title="Wifi-Data Magnitude and Frequency")
fig.show()

#Save to image
fig.write_image(f"/home/other/Wi-Fi-Analizer-Robot/Retrieve_Wi-Fi_Data/Pictures/fig{figIndex}.jpeg")
figIndex+=1
```
