```python
import pandas as pd
import plotly.io as pio
import plotly.express as px

pio.renderers.default = 'notebook_connected'
```

```python
path = "wifi.json"
wifiDf = pd.read_json(path)
wifiDf.head()
```

```python
wifiDf['Channel'] = wifiDf['Channel'].astype(int)
fig = px.scatter(wifiDf,x="Channel",y="Channel")
fig.show()
```
